```python
inputs = []

while True:
    user_input = input("Enter something (q to quit): ")
    if user_input == "q":
        break
    inputs.append(user_input)

for item in inputs:
    print(item)
```

```python
numbers = []
for i in range(10):
    num = int(input("Enter a number: "))
    numbers.append(num)

minimum = numbers[0]
maximum = numbers[0]

for n in numbers:
    if n < minimum:
        minimum = n
    if n > maximum:
        maximum = n

print("Minimum:", minimum)
print("Maximum:", maximum)
```

```python
import random

target = random.randint(1, 5000000)
found = False

for i in range(1, 1000001):
    if i == target:
        print("Number found")
        found = True
        break

if not found:
    print("Number not found")
```

```python
lst = [2, 3, 4, 5]
result = 1

for item in lst:
    result *= item

print(result)
```

```python
lst = [1, 2, 2, 3, 4, 4, 5]
unique = []

for item in lst:
    if item not in unique:
        unique.append(item)

print(unique)
```

```python
list1 = [1, 2, 3, 4]
list2 = [3, 4, 5, 6]

common = []

for item in list1:
    if item in list2:
        common.append(item)

print(common)
```

```python
list1 = [1, 2, 3, 4]
list2 = [3, 4, 5, 6]

not_common = []

for item in list1:
    if item not in list2:
        not_common.append(item)

for item in list2:
    if item not in list1:
        not_common.append(item)

print(not_common)
```

```python
lst = [1, 2, 3, 4, 5]

is_sorted = True

for i in range(len(lst) - 1):
    if lst[i] > lst[i + 1]:
        is_sorted = False
        break

print(is_sorted)
```
