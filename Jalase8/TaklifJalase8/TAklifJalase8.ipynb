```python
import os
import shutil


def toggle_folder():
    folder_name = input("Enter folder name: ").strip()
    if not folder_name:
        print("Invalid folder name.")
        return

    if os.path.exists(folder_name):
        if os.path.isdir(folder_name):
            shutil.rmtree(folder_name)
            print(f"Folder '{folder_name}' deleted.")
        else:
            print(f"'{folder_name}' is a file, not a folder.")
    else:
        os.makedirs(folder_name)
        print(f"Folder '{folder_name}' created.")


if __name__ == "__main__":
    toggle_folder()
```

```python
import os


def create_multiple_folders():
    base_path = input("Enter base path: ").strip()
    names_input = input("Enter folder names (comma separated): ").strip()

    if not names_input:
        print("No folder names given.")
        return

    os.makedirs(base_path, exist_ok=True)
    names = [n.strip() for n in names_input.split(",") if n.strip()]

    for name in names:
        full_path = os.path.join(base_path, name)
        os.makedirs(full_path, exist_ok=True)
        print(f"Created: {full_path}")


if __name__ == "__main__":
    create_multiple_folders()
```

```python
import os


def show_file_size():
    file_path = input("Enter file path: ").strip()

    if not os.path.isfile(file_path):
        print("File not found.")
        return

    size_bytes = os.path.getsize(file_path)
    size_kb = size_bytes / 1024
    size_mb = size_kb / 1024

    print(f"Size: {size_bytes} bytes ({size_kb:.2f} KB / {size_mb:.2f} MB)")


if __name__ == "__main__":
    show_file_size()
```

```python
import os
import time


def watch_folder(interval=5):
    folder_path = input("Enter folder path to watch: ").strip()

    if not os.path.isdir(folder_path):
        print("Folder not found.")
        return

    print(f"Watching '{folder_path}' every {interval} seconds. Press Ctrl+C to stop.")
    previous = set(os.listdir(folder_path))

    try:
        while True:
            time.sleep(interval)
            current = set(os.listdir(folder_path))

            added = current - previous
            removed = previous - current

            for name in added:
                print(f"[+] Added: {name}")
            for name in removed:
                print(f"[-] Removed: {name}")

            previous = current
    except KeyboardInterrupt:
        print("\nStopped watching.")


if __name__ == "__main__":
    watch_folder()
```

```python
import os
import shutil
from pathlib import Path

EXTENSION_MAP = {
    "image": {".jpg", ".jpeg", ".png", ".gif", ".bmp", ".svg", ".webp", ".tiff"},
    "audio": {".mp3", ".wav", ".aac", ".flac", ".ogg", ".m4a"},
    "video": {".mp4", ".mkv", ".avi", ".mov", ".wmv", ".flv", ".webm"},
    "document": {".pdf", ".doc", ".docx", ".txt", ".xls", ".xlsx", ".ppt", ".pptx", ".csv", ".md"},
}


def get_category(extension):
    extension = extension.lower()
    for category, extensions in EXTENSION_MAP.items():
        if extension in extensions:
            return category
    return "other"


def sort_files_by_type():
    folder_path = input("Enter folder path to sort: ").strip()

    if not os.path.isdir(folder_path):
        print("Folder not found.")
        return

    for item in os.listdir(folder_path):
        item_path = os.path.join(folder_path, item)

        if not os.path.isfile(item_path):
            continue

        ext = Path(item).suffix
        category = get_category(ext)

        target_dir = os.path.join(folder_path, category)
        os.makedirs(target_dir, exist_ok=True)

        target_path = os.path.join(target_dir, item)

        if os.path.exists(target_path):
            base = Path(item).stem
            target_path = os.path.join(target_dir, f"{base}_copy{ext}")

        shutil.move(item_path, target_path)
        print(f"Moved '{item}' -> {category}/")


if __name__ == "__main__":
    sort_files_by_type()
```
