```python
def my_sum(*args):
    return sum(args)
```

```python
my_sum(5,6,4,7,8,9,2)
```

```python
def my_sum(*args, **kwargs):
    print(kwargs)
    return sum(args)+ sum(kwargs.values())
```

```python
my_sum(2,3,4,5,6,7,8,a=40,b=30)
```

```python
def r_sum(n):
    if n <= 0:
        return 0
    return n + r_sum(n-1)
```

```python
r_sum(4)
```
