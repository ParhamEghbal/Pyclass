```python
def avglength(sentence):
    words = sentence.split()
    total_length = sum(len(w) for w in words)
    return total_length / len(words)
```

```python
def addString(num1, num2):
    return int(num1) + int(num2)
```

```python
def movSefr(lst):
    result = [x for x in lst if x != 0]
    zeros = [0] * lst.count(0)
    return result + zeros
```

```python
def fillNone(lst):
    result = []
    prev = 0

    for item in lst:
        if item is None:
            result.append(prev)
        else:
            result.append(item)
            prev = item
    return result
```
