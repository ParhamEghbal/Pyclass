```python
def reverse_list(lst):
    return lst[::-1]
print(reverse_list([1,2,3,4]))
```

```python
def max_mixed(lst):
    max_value = None

    for item in lst:
        if isinstance(item, str):
            value = len(item)
        else:
            value = item

        if max_value is None or value > max_value[1]:
            max_value = (item, value)

    return max_value[0]
print(max_mixed([10 , "hi" , 99 , 54 ]))
```

```python
def check_speed(speed):
    if speed < 70:
        return "good"

    points = (speed - 70) // 5

    if points > 12:
        print("Govahi Name At Batele")

        return points

    return points
print(check_speed(100))
```

```python
import math

def euclidean_distance(a, b):
    return math.sqrt((b[0] - a[0])**2 + (b[1] - a[1])**2)
print(euclidean_distance((0 , 0) , (3 , 4)))
```

```python
import math

def sort_by_distance(points):
    return sorted(points, key=lambda p: math.sqrt(p[0]**2 + p[1]**2))

print(sort_by_distance([(1,1) , (5 , 5) , (0 , 2)]))
```
