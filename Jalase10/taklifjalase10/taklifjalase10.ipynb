```python
import winsound
import time

class IceCream:
    def __init__(self, name, weight, price, is_pair=False, sound=None):
        self.name = name
        self.weight = weight
        self.price = price
        self.is_pair = is_pair
        self.sound = sound
        self.stock = 0

    def play_sound(self):
        if self.sound:
            try:
                winsound.Beep(self.sound, 500)
            except:
                print("Playing music...")
            time.sleep(0.5)


class Shop:
    def __init__(self):
        self.capacity = 200000
        self.total_weight = 0
        self.products = {
            "1": IceCream("Chocolate", 100, 10),
            "2": IceCream("Fruit", 55, 7),
            "3": IceCream("Musical", 200, 15, is_pair=True, sound=1000),
            "4": IceCream("Ice", 40, 3)
        }
        self.total_sales = 0

    def show_menu(self):
        print("\n--- Menu ---")
        for key, item in self.products.items():
            pair_text = " (Pair)" if item.is_pair else ""
            print(f"{key}. {item.name} | Weight: {item.weight}g | Price: {item.price}{pair_text}")
        print("5. Buy")
        print("6. Add Stock")
        print("7. Show Stock")
        print("8. Show Sales")
        print("9. Exit")

    def add_stock(self):
        print("\n--- Add Stock ---")
        for key, item in self.products.items():
            try:
                qty = int(input(f"Enter quantity for {item.name}: "))
                if qty < 0:
                    print("Invalid quantity.")
                    continue
                new_weight = qty * item.weight
                if self.total_weight + new_weight > self.capacity:
                    print(f"Not enough space for {item.name}.")
                    continue
                item.stock += qty
                self.total_weight += new_weight
            except ValueError:
                print("Invalid input.")

    def buy(self):
        print("\n--- Buy ---")
        cart = []
        total_price = 0
        total_added_weight = 0

        for key, item in self.products.items():
            try:
                qty = int(input(f"Enter quantity for {item.name} (0 to skip): "))
                if qty <= 0:
                    continue
                
                if item.is_pair:
                    if qty % 2 != 0:
                        print(f"{item.name} must be bought in pairs. Adjusting to even number.")
                        qty += 1
                
                if qty > item.stock:
                    print(f"Not enough stock for {item.name}. Available: {item.stock}")
                    continue
                
                added_weight = qty * item.weight
                if self.total_weight + added_weight > self.capacity:
                    print(f"Cannot add {item.name}. Warehouse capacity exceeded.")
                    continue
                
                item.stock -= qty
                self.total_weight -= added_weight
                total_added_weight += added_weight
                
                cost = qty * item.price
                total_price += cost
                cart.append((item, qty, cost))
                
                if item.sound:
                    print(f"Playing music for {item.name}...")
                    item.play_sound()
                    
            except ValueError:
                print("Invalid input.")
        
        if not cart:
            print("Cart is empty.")
            return
        
        print("\n--- Receipt ---")
        for item, qty, cost in cart:
            print(f"{item.name} x {qty} = {cost}")
        print(f"Total: {total_price}")
        
        self.total_sales += total_price
        print("Purchase successful.")

    def show_stock(self):
        print("\n--- Stock ---")
        for key, item in self.products.items():
            print(f"{item.name}: {item.stock} units | Weight: {item.stock * item.weight}g")
        print(f"Total Weight: {self.total_weight}g / {self.capacity}g")

    def show_sales(self):
        print(f"\nTotal Sales: {self.total_sales}")
        print(f"Total Profit (20%): {self.total_sales * 0.2}")

    def run(self):
        while True:
            self.show_menu()
            choice = input("Choose an option: ")
            
            if choice == "1" or choice == "2" or choice == "3" or choice == "4":
                item = self.products[choice]
                print(f"{item.name} | Stock: {item.stock}")
            elif choice == "5":
                self.buy()
            elif choice == "6":
                self.add_stock()
            elif choice == "7":
                self.show_stock()
            elif choice == "8":
                self.show_sales()
            elif choice == "9":
                print("Goodbye!")
                break
            else:
                print("Invalid option.")


if __name__ == "__main__":
    shop = Shop()
    shop.run()
```
