```python
class ball :

    def __init__(self,color,number,x,y):
        self.color = color
        self.number = number
        self.move(x,y)

    def move(self,x,y):
        self.x = x
        self.y = y

    def reset(self):
        self.move(0,0)

    def dist(self,other):
        return ((self.x - other.x)**2 + (self.y - other.y)**2)**.5

    def is_collide(self,other):
        return self.dist(other) <= 0.5

    def __repr__(self):
        return f'ball num{self.number},ball color:{self.color}'
    
```

```python
b1 = ball('red','1',21.5,15.6)
b2 = ball('blue','2',25.5,35.6)
b3 = ball('yelow','3',26,36)
```
